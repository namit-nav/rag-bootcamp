## CSV & EXCEL FILES - Structured Data

```python
import pandas as pd
import os
import openpyxl
```

```python
os.makedirs("data/structured_files", exist_ok=True)
```

```python
# Create a sample data
data = {
    'Product': ['Laptop', 'Smartphone', 'Tablet'],
    'Price': [999.99, 499.99, 299.99],
    'Stock': [10, 50, 30],
    'Category': ['Electronics', 'Electronics', 'Electronics'],
    'Description': ['A high-end laptop', 'A powerful smartphone', 'A versatile tablet']
}

# Save as CSV
csv_file_path = "data/structured_files/products.csv"
pd.DataFrame(data).to_csv(csv_file_path, index=False)
```

```python
# Save as Excel
excel_file_path = "data/structured_files/products.xlsx"
pd.DataFrame(data).to_excel(excel_file_path, index=False)
```

### CSV Processing

```python
from langchain_community.document_loaders import CSVLoader, UnstructuredExcelLoader
```

```python
# Method 1: Using CSVLoader
print("Loading CSV file using CSVLoader...")
csv_loader = CSVLoader(
    file_path=csv_file_path,
    encoding="utf-8",
    csv_args={
        "delimiter": ",",
        'quotechar': '"',
    }
)

csv_documents = csv_loader.load()
print(f"Loaded {len(csv_documents)} documents from CSV file.")
print("\nFirst document from CSV:")
print(csv_documents[0].page_content)
```

```python
# Method 2: Using UnstructuredExcelLoader
print("\nLoading Excel file using UnstructuredExcelLoader...")
excel_loader = UnstructuredExcelLoader(
    file_path=excel_file_path,
    encoding="utf-8",
    excel_args={
        "sheet_name": "Sheet1"
    }
)

excel_documents = excel_loader.load()
print(f"Loaded {len(excel_documents)} documents from Excel file.")
print("\nFirst document from Excel:")
print(excel_documents[0].page_content)
```
