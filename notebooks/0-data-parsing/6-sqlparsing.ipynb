### SQL Databases

```python
# sample SQLite database
import sqlite3
import os
os.makedirs("data/databases/sqlite_db", exist_ok=True)
```

```python
# create a sample SQLite database and table
db_path = "data/databases/sqlite_db/products.db"

conn = sqlite3.connect(db_path)
cursor = conn.cursor()
cursor.execute('''CREATE TABLE IF NOT EXISTS products (
                    id INTEGER PRIMARY KEY,
                    name TEXT,
                    price REAL,
                    description TEXT
                )''')
```

```python
cursor.execute('''CREATE TABLE IF NOT EXISTS customers (
                    id INTEGER PRIMARY KEY,
                    name TEXT,
                    email TEXT
                )''')
```

```python
# Insert sample data into products table
products_data = [
    (1, "Laptop", 999.99, "A high-performance laptop for work and play."),
    (2, "Smartphone", 499.99, "A sleek smartphone with a great camera."),
    (3, "Headphones", 199.99, "Noise-cancelling headphones for immersive sound.")
]

cursor.executemany('INSERT INTO products VALUES (?, ?, ?, ?)', products_data)
```

```python
# Insert sample data into customers table
customers_data = [
    (1, "Alice Smith", "alice.smith@example.com"),
    (2, "Bob Johnson", "bob.johnson@example.com"),
    (3, "Charlie Brown", "charlie.brown@example.com")
]

cursor.executemany('INSERT INTO customers VALUES (?, ?, ?)', customers_data)
```

```python
cursor.execute('''SELECT * FROM products''')
```

```python
conn.commit()
conn.close()
```

### Database content extraction

```python
from langchain_community.utilities import SQLDatabase
```

```python
# Method 1: Using SQLDatabase
db = SQLDatabase.from_uri(f"sqlite:///{db_path}")

# get the database info
print(f"Tables: {db.get_usable_table_names()}")
print(f"Schema for tables: {db.get_table_info()}")

```

```python
# Method 2: Using SQLDatabase with custom settings
db_custom = SQLDatabase.from_uri(
    f"sqlite:///{db_path}",
    include_tables=["products"],  # Only include the 'products' table
    sample_rows_in_table_info=2  # Limit the number of sample rows in table info
)
# get the database info for the custom settings
print(f"Custom Tables: {db_custom.get_usable_table_names()}")
print(f"Custom Schema for tables: {db_custom.get_table_info()}")
```
